```python
from pathlib import Path
import shutil
from yt_dlp import YoutubeDL

def download_audio(link, path="audio", cookies_from_browser=None):
    output_dir = Path(path)
    output_dir.mkdir(parents=True, exist_ok=True)
    node_path = shutil.which("node")

    ydl_opts = {
        "format": "bestaudio/best",
        "postprocessors": [{
            "key": "FFmpegExtractAudio",
            "preferredcodec": "wav",
        }],
        "outtmpl": str(output_dir / "%(title)s.%(ext)s"),
        "noplaylist": True,
        "remote_components": ["ejs:github"],
    }

    if node_path:
        ydl_opts["js_runtimes"] = {"node": {"path": node_path}}
    if cookies_from_browser:
        ydl_opts["cookiesfrombrowser"] = (cookies_from_browser,)

    with YoutubeDL(ydl_opts) as ydl:
        ydl.download([link])

    print(f"Audio downloaded from {link} and saved to {output_dir}")
```

```python
from pathlib import Path
import re

for directory in (Path.cwd(), *Path.cwd().parents):
    podcasts_file = directory / "data" / "podcasts.md"
    if podcasts_file.is_file():
        break
else:
    raise FileNotFoundError("Could not find data/podcasts.md")

links = re.findall(
    r"https?://[^\s)>]+",
    podcasts_file.read_text(encoding="utf-8"),
)

print(links)
```

```python
cookies_from_browser = "chrome"  # Set to "chrome" or "firefox" if yt-dlp requests sign-in
failed_downloads = []

for index, link in enumerate(links, start=1):
    try:
        print(f"({index}/{len(links)}) Downloading {link}")
        download_audio(link, cookies_from_browser=cookies_from_browser)
    except Exception as error:
        failed_downloads.append((link, str(error)))
        print(f"Failed: {error}")

print(f"Finished: {len(links) - len(failed_downloads)} succeeded, {len(failed_downloads)} failed.")
```
