```python
from yt_dlp import YoutubeDL

def download_audio(link, path="audio"):
    ydl_opts = {
        'format': 'bestaudio/best',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'wav',
        }],
        'outtmpl': f'{path}/%(title)s.%(ext)s'
    }

    with YoutubeDL(ydl_opts) as ydl:
        ydl.download([link])

    print(f"Audio downloaded from {link} and saved to {path}")
```

```python
from pathlib import Path
import re

for directory in (Path.cwd(), *Path.cwd().parents):
    podcasts_file = directory / "data" / "podcasts.md"
    if podcasts_file.is_file():
        break
else:
    raise FileNotFoundError("Could not find data/podcasts.md")

links = re.findall(
    r"https?://[^\s)>]+",
    podcasts_file.read_text(encoding="utf-8"),
)

print(links)
```

```python
for link in links:
    download_audio(link)
```
