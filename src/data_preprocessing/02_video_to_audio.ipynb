# Step 2 (optional): Convert videos to audio

Converts every video in `data/videos/` (`.mp4`, `.mkv`, `.mov`, `.webm`) to a WAV file with the same name in `data/raw_audios/`. Videos that were already converted are skipped.

Needs `ffmpeg` on your PATH. Skip this step if you have no video files.

```python
import subprocess
from pathlib import Path

# Find the project folder (works from any notebook folder)
project_root = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "pyproject.toml").is_file() and (path / "src").is_dir()
)
video_dir = project_root / "data" / "videos"
output_dir = project_root / "data" / "raw_audios"
output_dir.mkdir(parents=True, exist_ok=True)

VIDEO_TYPES = {".mp4", ".mkv", ".mov", ".webm"}
videos = sorted(p for p in video_dir.glob("*") if p.suffix.lower() in VIDEO_TYPES)
print(f"Found {len(videos)} videos in {video_dir.relative_to(project_root)}")
```

## Convert

```python
for video in videos:
    wav_file = output_dir / f"{video.stem}.wav"
    if wav_file.exists():
        print(f"Already converted: {wav_file.name}")
        continue
    # -vn drops the picture; pcm_s16le is plain 16-bit WAV
    subprocess.run(
        ["ffmpeg", "-loglevel", "error", "-i", str(video), "-vn", "-c:a", "pcm_s16le", str(wav_file)],
        check=True,
    )
    print(f"Saved: {wav_file.name}")
```
